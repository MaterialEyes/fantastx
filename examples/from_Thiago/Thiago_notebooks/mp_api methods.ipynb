```python
from mp_api.client import MPRester
from pymatgen.core.structure import Structure
import matplotlib.pyplot as plt 
```

```python
mpr = MPRester("RKRcS7A9ek75SYfPCFYzXiaFvNQdba7L")
```

```python
mpr.summary.search?
```

```python
cubic_crys_sys = mpr.summary.search(crystal_system='Cubic', energy_above_hull=[0.0, 0.001], )
```

```python
first3 = cubic_crys_sys[:3]
```

```python
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
SGA = SpacegroupAnalyzer(first3[1].structure)
conventional_structure = SGA.get_conventional_standard_structure()
```
