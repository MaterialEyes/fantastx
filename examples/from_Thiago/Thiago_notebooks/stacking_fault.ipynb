```python
# Read the initial cluster system from the CIF file
cluster = read("Pt_6layer.cif")

# Select 2 3-atom pairs in adjacent atomic planes that will serve as reference for the shear displacement
plane1_atom1 = [25.391, 41.795, 37.774]
plane1_atom2 = [25.391, 41.795, 32.226]
plane1_atom3 = [28.594, 32.735, 37.774]

plane2_atom1 = [26.992, 44.060, 37.774]
plane2_atom2 = [27.793, 41.795, 30.839]
plane2_atom3 = [30.996, 32.735, 36.387]
```

```python
cluster = read("Pt_6layer.cif")
# Input displacement vector magnitude
displacement_magnitude = 1.0
# Calculate the stacking fault plane normal vector for each set of planes
plane1_hkl = np.cross(np.array(plane1_atom2) - np.array(plane1_atom1),
                      np.array(plane1_atom3) - np.array(plane1_atom1))
plane2_hkl = np.cross(np.array(plane2_atom2) - np.array(plane2_atom1),
                      np.array(plane2_atom3) - np.array(plane2_atom1))

# Normalize the displacement direction vectors
displacement_direction = np.cross(plane1_hkl, plane2_hkl)
displacement_direction /= np.linalg.norm(displacement_direction)

# User input: Choose the direction for displacement
# 1 for right, -1 for left, 0 for no displacement
displacement_flag = 1

# User input: Choose the direction for displacement
# 2 for up, -2 for down, 0 for no displacement
displacement_flag_2 = 2

# User input: Provide the atomic positions of three atoms for each set of parallel planes
# Format: [x, y, z]

# Calculate the stacking fault plane normal vector for the first set of planes
plane2_hkl = np.cross(np.array(plane2_atom2) - np.array(plane2_atom1),
                      np.array(plane2_atom3) - np.array(plane2_atom1))

# Normalize the stacking fault plane normal vector
plane1_hkl /= np.linalg.norm(plane2_hkl)

# Find the distance from each atom to the plane defined by the first set of atoms
for atom in cluster:
    distance_to_plane2 = np.dot(atom.position - plane2_atom1, plane2_hkl)

     # Calculate the distance of the atom to the planes
    distance_to_plane1 = np.abs(np.dot(atom.position - plane1_atom1, plane1_hkl) / np.linalg.norm(plane1_hkl))
    #distance_to_plane2 = np.abs(np.dot(atom.position - plane2_atom1, plane2_hkl) / np.linalg.norm(plane2_hkl))
    
    # Calculate the displacement vectors for the atom based on the direction flags
    displacement_vector = displacement_flag * displacement_magnitude * displacement_direction
    displacement_vector_2 = displacement_flag_2 * displacement_magnitude * np.array([0, 0, 1])

    # Print the coordinates of atoms to the left of the displaced plane
    if distance_to_plane2 > 0.1:
        #print("Atom coordinates:",atom)
        if distance_to_plane1 != 0 and displacement_flag_2 != 0 :
            atom.position += displacement_vector
        if displacement_flag_2 != 0 :
            atom.position += displacement_vector_2

view(cluster)

#Figure out a wy to move the adjacent atoms to the displaced plane
```
